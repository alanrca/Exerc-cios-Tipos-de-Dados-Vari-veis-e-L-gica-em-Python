2️⃣ Análise de Receita

Crie duas variáveis:

Receita do mês atual

Receita do mês anterior

Calcule:

Diferença de receita

Percentual de crescimento ou queda

Imprima o resultado formatado.

```python
# Crie duas variáveis:Receita do mês atual e Receita do mês anterior
# Calcule: Diferença de receita e Percentual de crescimento ou queda 
# Imprima o resultado formatado.
rec_mes_atual = 2000.00
rec_mes_anterior = 1000.00

diferenca_receita = rec_mes_atual - rec_mes_anterior
perc_crescimento = (diferenca_receita / rec_mes_anterior) * 100

print(
    f"A receita do mês anterior foi R$ {rec_mes_anterior:.2f}, "
    f"e a do mês atual foi R$ {rec_mes_atual:.2f}, "
    f"gerando uma diferença de R$ {diferenca_receita:.2f} "
    f"({perc_crescimento:.2f}% de crescimento)."
)
```

1️⃣1️⃣ Lista de Clientes (Desafio)

Crie uma lista com 3 dicionários, onde cada dicionário representa um cliente contendo:

* Nome

* Faturamento

* Ativo (True ou False)

Depois:

1. Calcule o faturamento total apenas dos clientes ativos

2. Conte quantos clientes estão inativos

📌 Esse exercício simula muito o que fazemos com pandas depois.

```python
#Criar a lista de clientes
clientes = [
    {"nome": "Maria", "faturamento": 12000.00, "ativo": True},
    {"nome": "João", "faturamento": 8000.00, "ativo": False},
    {"nome": "Ana", "faturamento": 15000.00, "ativo": True}
]

# Faturamento total dos ativos
faturamento_total = 0
inativos = 0

for cliente in clientes:
    if cliente["ativo"]:
        faturamento_total += cliente["faturamento"]
    else:
        inativos += 1

print(f"Faturamento total dos clientes ativos: R$ {faturamento_total:.2f}")
print(f"Quantidade de clientes inativos: {inativos}")

```

🔹 BLOCO 7 – Lógica Combinada (Desafio Final)

1️⃣2️⃣ Score de Cliente

Para cada cliente, aplique a regra:

Score A → faturamento ≥ 20000 e ativo

Score B → faturamento ≥ 10000 e ativo

Score C → faturamento < 10000 e ativo

Score D → inativo

Imprima o nome do cliente e seu score.

```python
#Criar lista de clientes:
clientes = [
    {"nome": "Maria", "faturamento": 20000.00, "ativo": True},
    {"nome": "João", "faturamento": 10000.00, "ativo": False},
    {"nome": "Ana", "faturamento": 10000.00, "ativo": True},
    {"nome": "Joanalina", "faturamento": 5000.00, "ativo": True}
]
for cliente in clientes:
    if cliente["ativo"] and cliente["faturamento"] >= 20000:
        score = "A"
    elif cliente["ativo"] and cliente["faturamento"] >= 10000:
        score = "B"
    elif cliente["ativo"] and cliente["faturamento"] < 10000:
        score = "C"
    else:
        score = "D"
    
    print(f"Cliente: {cliente['nome']} → Score {score}")
```
